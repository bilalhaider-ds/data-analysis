# How to create new columns

```python
df["new column"] = 5
df
```

How to create a column with serial number (numpy function)

```python
df["SN"] = np.arange(1,1001)
df
```

How to create a column with random numbers

```python
df["Random Numbers"] = np.random.randint(5, 100,df.shape[0])
df
```

```python
df["Random Numbers"] = np.random.randint(9, 99,df.shape[0])
df
```

```python
df["New Price"] = df["Price per Unit"] / 2
df
```
