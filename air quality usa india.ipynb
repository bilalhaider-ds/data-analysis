```python
import pandas as pd

udf = pd.read_csv("usa air quality.csv")
udf
```

```python
idf = pd.read_csv("india air quality.csv")
idf
```

comparative analysis steps

```python
# comparing with mean

idf_mean = idf["AQI"].mean()
udf_mean = udf["AQI"].mean()
idf_mean, udf_mean
```

```python
# comparing with median

idf_median = idf["AQI"].median()
udf_median = udf["AQI"].median()
idf_median, udf_median
```

```python
# comparing with mode
idf_mode = idf["AQI"].mode()[0]
udf_mode = udf["AQI"].mode()[0]
idf_mode, udf_mode
```

```python
# comparing with max and min

idf_max = idf["AQI"].max()
udf_max = udf["AQI"].max()
idf_min = idf["AQI"].min()
udf_min = udf["AQI"].min()
idf_max, udf_max, idf_min, udf_min
```

```python
idf.loc[idf["AQI"] == idf_max]
```

```python
udf.loc[udf["AQI"] == udf_max]
```

```python
idf.loc[idf["AQI"] == idf_min]
```

```python
udf.loc[udf["AQI"] == udf_min]
```

```python
#top 5 most concerning India and USA cities

idf_top5 = idf.sort_values(by="AQI", ascending=False).head(5)
udf_top5 = udf.sort_values(by="AQI", ascending=False).head(5)
idf_top5, udf_top5
```
