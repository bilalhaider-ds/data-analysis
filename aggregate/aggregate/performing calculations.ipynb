```python
import pandas as pd

data = pd.read_csv(r'/home/bilal-haider/console_flare/score.csv')
data
```

```python
# single column sum
data["Math"].sum()
```

```python
# multiple column sum
data[["Math", "Science"]].sum()
```

```python
# use iloc to get the sum of a specific row
data.iloc[0:, 1:6].sum()
```

```python
# every rows sum
data.iloc[:, 1:6].sum(axis=1)
```

```python
#storing sum in a new column
data["Total"] = data.iloc[:, 1:6].sum(axis=1)
data
```

```python
# max value in a  single column
data["Math"].max()
```

```python
# max value in multiple columns
data[["Math", "Science"]].max()
```

```python
# max value of all subjects   
data.iloc[:, 1:6].max()
```

```python
#maximun score of each student
data["Maximum Score"] = data.iloc[:, 1:6].max(axis=1)
data["Maximum Score"]
```

```python
# min value in a  single column
data["Math"].min()
```

```python
# min value in multiple columns
data[["Math", "Science"]].min()
```

```python
# min value of all subjects
data.iloc[:, 1:6].min()
```

```python
data["Minimum Score"] = data.iloc[:, 1:6].min(axis=1)
data["Minimum Score"]
```
